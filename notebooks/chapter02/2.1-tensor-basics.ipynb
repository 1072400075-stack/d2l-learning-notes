## 1. 创建张量

标量是 0 维，向量是 1 维，矩阵是 2 维。更高维的数据也统一称为张量。

```python
scalar = torch.tensor(7)
vector = torch.tensor([1.0, 2.0, 3.0])
matrix = torch.tensor([[1.0, 2.0], [3.0, 4.0]])

print('标量:', scalar, 'shape =', scalar.shape)
print('向量:', vector, 'shape =', vector.shape)
print('矩阵:\n', matrix, 'shape =', matrix.shape)
```

## 2. 形状与 reshape

`reshape` 可以改变排列方式，但不能改变元素总数。

```python
x = torch.arange(12)
X = x.reshape(3, 4)

print('原张量:', x)
print('原形状:', x.shape)
print('重塑后:\n', X)
print('新形状:', X.shape)
print('元素数量:', X.numel())
```

## 4. 运算与广播

广播会在形状兼容时自动扩展长度为 1 的维度。

```python
a = torch.arange(3).reshape(3, 1)
b = torch.arange(2).reshape(1, 2)

print('a 的形状:', a.shape)
print('b 的形状:', b.shape)
print('a + b =\n', a + b)
print('结果形状:', (a + b).shape)
```

## 5. 在 GPU 上计算

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
gpu_matrix = torch.randn(1000, 1000, device=device)
result = gpu_matrix @ gpu_matrix

print('计算设备:', result.device)
print('结果形状:', result.shape)
```
